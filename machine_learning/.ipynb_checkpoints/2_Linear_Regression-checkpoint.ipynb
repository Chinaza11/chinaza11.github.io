### Import packages

```python
import os, numpy as np, pandas as pd
import matplotlib.pyplot as plt
from itertools import combinations
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
```

```python
# =============================================================================
#  My Linear Regression Code
# =============================================================================

class LinearRegressionAnalysis:
    def __init__(self, data):
        self.data = data
        self.correlated_pairs = []
        self.plots_list = []  

    def linear_regression(self, X, Y):
        X_mean = np.mean(X)
        Y_mean = np.mean(Y)
        
        covariance_x_y = np.sum((X - X_mean) * (Y - Y_mean))
        variance_x = np.sum((X - X_mean) ** 2)
        
        self.theta1 = covariance_x_y / variance_x
        self.theta0 = Y_mean - self.theta1 * X_mean
        return self.theta1, self.theta0

    def correlation_coefficient(self, x, y):
        x_mean = np.mean(x)
        y_mean = np.mean(y)
        numerator = np.sum((x - x_mean) * (y - y_mean))
        denominator = np.sqrt(np.sum((x - x_mean) ** 2) * np.sum((y - y_mean) ** 2))
        return numerator / denominator

    # Find all pairs of variables with correlation above the threshold
    def find_correlated_pairs(self, threshold=0.8):
        variable_pairs = list(combinations(self.data.columns, 2))
        for var1, var2 in variable_pairs:
            x = self.data[var1]
            y = self.data[var2]
            corr = self.correlation_coefficient(x, y)
            if corr**2 > threshold:
                self.correlated_pairs.append((var1, var2, corr))
        print(f"Found {len(self.correlated_pairs)} correlated pairs.")
        for i in self.correlated_pairs:
            print(i)
        return self.correlated_pairs

    # Plot scatter plots for correlated pairs in groups of 16 on a 4x4 grid
    def plot_correlated_pairs(self):
        max_plots_per_figure = 16
        num_figures = (len(self.correlated_pairs) + max_plots_per_figure - 1) // max_plots_per_figure

        for fig_idx in range(num_figures):
            fig, axes = plt.subplots(4, 4, figsize=(16, 16))
            axes = axes.flatten()
            start_idx = fig_idx * max_plots_per_figure
            end_idx = min(start_idx + max_plots_per_figure, len(self.correlated_pairs))

            for i, (var1, var2, corr) in enumerate(self.correlated_pairs[start_idx:end_idx]):
                ax = axes[i]
                x = self.data[var1]
                y = self.data[var2]

                # Perform linear regression
                slope, intercept = self.linear_regression(x, y)
                regression_line = slope * x + intercept

                # Scatter plot
                ax.scatter(x, y, alpha=0.7, label=f'{var1} vs {var2}')
                ax.plot(x, regression_line, color='red')

                # Annotate with R2
                ax.set_title(f'{var1} vs {var2} (r = {corr:.2f})')
                ax.set_xlabel(var1)
                ax.set_ylabel(var2)
                ax.grid(True)

            # Hide unused subplots
            for j in range(i + 1, len(axes)):
                axes[j].axis('off')

            plt.tight_layout()
            self.plots_list.append(fig) # Save figure to the list
            plt.show()

    def save_plots(self, save_paths):
        for fig, path in zip(self.plots_list[:len(save_paths)], save_paths):
            fig.savefig(path, format='png', dpi=300, bbox_inches='tight')
        print(f"Saved {len(save_paths)} plots.")
```

```python
# =============================================================================
# Sklearn
# =============================================================================

def find_correlated_pairs(data, threshold=0.8):
    variable_pairs = list(combinations(data.columns, 2))
    correlated_pairs = []
    for var1, var2 in variable_pairs:
        x = data[var1]
        y = data[var2]
        corr = r2_score(x, y)
        if corr > threshold:
            correlated_pairs.append((var1, var2, corr))
    print(f"Found {len(correlated_pairs)} correlated pairs.")
    for i in correlated_pairs:
        print(i)
    return correlated_pairs
    
sklearn_result = find_correlated_pairs(transposed_data)

# Plot scatter plots for correlated pairs in groups of 16 on a 4x4 grid
def plot_correlated_pairs(data, correlation_result_list):
    plots_list = []
    max_plots_per_figure = 16
    num_figures = (len(correlation_result_list) + max_plots_per_figure - 1) // max_plots_per_figure

    for fig_idx in range(num_figures):
        fig, axes = plt.subplots(4, 4, figsize=(16, 16))
        axes = axes.flatten()
        start_idx = fig_idx * max_plots_per_figure
        end_idx = min(start_idx + max_plots_per_figure, len(correlation_result_list))

        for i, (var1, var2, corr) in enumerate(correlation_result_list[start_idx:end_idx]):
            ax = axes[i]
            x = data[[var1]]
            y = data[var2]
            
            model = LinearRegression()
            model.fit(x,y)
            y_pred = model.predict(x)

            # Scatter plot
            ax.scatter(x, y, alpha=0.7, label=f'{var1} vs {var2}')
            ax.plot(x, y_pred, color='red')

            # Annotate with correlation coefficient
            ax.set_title(f'{var1} vs {var2} (r = {corr:.2f})')
            ax.set_xlabel(var1)
            ax.set_ylabel(var2)
            ax.grid(True)

        # Hide unused subplots
        for j in range(i + 1, len(axes)):
            axes[j].axis('off')

        plt.tight_layout()
        plots_list.append(fig)
        plt.show()
    return plots_list

sklearn_plots_list = plot_correlated_pairs(transposed_data, sklearn_result)
```

```python
def save_plots(plots_list, save_paths):
    for fig, path in zip(plots_list[:len(save_paths)], save_paths):
        fig.savefig(path, format='png', dpi=300, bbox_inches='tight')
    print(f"Saved {len(save_paths)} plots.")

# Save the first two plots
save_plots(sklearn_plots_list, ["sklearn_LR_plot1.png", "sklearn_LR_plot2.png"])
```
